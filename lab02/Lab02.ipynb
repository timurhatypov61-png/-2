## Задание 6.2 — описательная статистика

Через store_df.describe() получена сводка по числовым столбцам.

- order_sum: 15 наблюдений, mean = 5000, std = 11096, медиана = 2100,
  max = 45000.
- delivery_time: 12 наблюдений (3 пропуска не учитываются),
  mean = 39.58, медиана = 37.5, max = 70.
- rating: 15 наблюдений, mean = 3.8, диапазон 1–5.

Ответ на контрольный вопрос: столбец order_sum имеет max = 45000,

```python
mean_delivery = store_df["delivery_time"].mean() #среднее
median_delivery = store_df["delivery_time"].median() #медиана

print("Среднее:", mean_delivery)
print("Медиана:", median_delivery)
```

## Задание 6.3 — среднее и медиана времени доставки

Среднее время доставки = 39.58 минут, медиана = 37.5 минут.
Пропуски (NaN) автоматически игнорируются методами mean() и median().

Ответ на контрольный вопрос:  
Среднее и медиана отличаются, потому что считаются по-разному.
Среднее учитывает каждое значение и чувствительно к выбросам.
Медиана — это середина отсортированного ряда, она устойчива
к экстремальным значениям. В наших данных есть выброс 70 минут,
который тянет среднее вверх, поэтому среднее (39.58) немного
больше медианы (37.5).

```python
print(store_df["rating"].value_counts()) #сколько раз встречается каждое уникальное значение
```

## Задание 6.4 — распределение рейтингов

Через store_df["rating"].value_counts() получено распределение оценок:

- 4 → 4 заказа

Ответ на контрольный вопрос: наиболее часто встречается оценка 5 —

```python
store_df["is_high_value"] = store_df["order_sum"] > 3000 #создание признака
print(store_df[["order_id", "order_sum", "is_high_value"]])
```

## Задание 7.1 — производный признак is_high_value

Через store_df["is_high_value"] = store_df["order_sum"] > 3000
создан булев столбец. True — сумма больше 3000, False — меньше
или равна 3000.

Заказов с is_high_value = True: 4 (order_id 1004, 1008, 1011, 1015).

```python
channel_stats = store_df.groupby("channel")["order_sum"].mean() #группировка по каналу продаж
print(channel_stats)
```

## Задание 7.2 — группировка по каналу продаж

Через groupby("channel")["order_sum"].mean() рассчитана средняя
сумма заказа для каждого канала:
- app: 1800.00
- marketplace: 13775.00
- site: 1816.67

Средняя сумма marketplace в 7 раз выше остальных каналов.
Причина — заказ с суммой 45000, который попал в этот канал
и сильно завысил среднее. Без него средняя по marketplace
была бы около 3367.

```python
channel_stats = channel_stats.sort_values(ascending=False) #сортировка результата в порядĸе убывания
print(channel_stats)
```

## Задание 7.3 — сортировка

Через sort_values(ascending=False) результат отсортирован
по убыванию средней суммы заказа:
1. marketplace — 13775.00
2. site — 1816.67
3. app — 1800.00
