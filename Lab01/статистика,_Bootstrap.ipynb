```python
orders = np.array([
    [35, 1200, 2],
    [42, 3500, 5],
    [28, 800, 1],
    [51, 4200, 7],
    [39, 1900, 3],
    [63, 5600, 8],
    [31, 1500, 2],
    [45, 2800, 4]
]) #1.время доставки; 2.стоимость заказа; 3.количество товаров.
print(orders.shape)

print(orders[0]) #данные первого заказа
print(orders[:,0]) #время доставки всех заказов
print(orders[:,1]) #стоимость всех заказов
print(orders[5,2]) #количество товаров в шестом заказе

print(orders[2:]) #все заказы начиная с третьего
print(orders[:5, :2]) #первые пять заказов и первые два признака
```

## Вывод по заданию 1.2

Матрица orders имеет форму (8, 3): 8 строк — это 8 заказов,
3 столбца — время доставки, стоимость и количество товаров.

Индексация orders[i, j] позволяет получить конкретный элемент,
orders[i] — целую строку (заказ), orders[:, j] — целый столбец (признак).
Срезы orders[a:b, c:d] позволяют выбрать подмножество строк и столбцов.

orders.shape возвращает (8, 3), потому что в матрице 8 строк и 3 столбца —
это описывает структуру данных. orders.size возвращает 24, потому что
общее количество элементов = 8 × 3 = 24 — это описывает объём данных.

```python
orders = np.array([
    [35, 1200, 2],
    [42, 3500, 5],
    [28, 800, 1],
    [51, 4200, 7],
    [39, 1900, 3],
    [63, 5600, 8],
    [31, 1500, 2],
    [45, 2800, 4]
]) #1.время доставки; 2.стоимость заказа; 3.количество товаров.

delivery = orders[:, 0] #столбец времени доставки

delivery[0] = 1000 #Измените первый элемент delivery на 1000
print(orders) #Выведите orders
```

первая строка orders изменилась — вместо [35 1200 2] стало [1000 1200 2]. То есть изменение delivery повлияло на оригинал. Это и есть view.

```python
delivery[0] = 35 #Восстановите исходную матрицу orders
delivery_copy = orders[:, 0].copy() #Создайте независимую копию
delivery_copy[0] = 1000 #Измените первый элемент delivery_copy
print(orders) #Снова выведите orders
```

## Вывод по заданию 1.3

Срез orders[:, 0] без .copy() возвращает view — представление на исходные
данные. Изменение view меняет и оригинальную матрицу, потому что оба
объекта ссылаются на одну и ту же память.

Метод .copy() создаёт независимую копию. Изменения в копии не влияют
на исходный массив.

Разница важна при обработке реальных данных: случайное изменение view
может незаметно испортить исходный датасет. Например, аналитик выделил
столбец с ценами, чтобы временно их нормализовать, и случайно перезаписал
исходные значения. Дальше все отчёты строятся на искажённых данных, и
ошибку можно заметить не сразу.

```python
delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])

mask = delivery_time > 45 #булева маска для поиска заказов с временем доставки больше 45 минут
print(mask) #вывод маски
print(delivery_time[mask]) #результат применения маски
print(delivery_time[mask].size) #количество проблемных заказов
print(delivery_time[mask].mean()) #среднее время доставки для этой группы
```

Разница между > 45 и >= 45: строгое неравенство не включает заказ с

Анализ комбинации признаков полезнее, чем анализ каждого по отдельности,

и крупные заказы тоже. Но их пересечение может выявить проблему: склад

При наличии редких экстремальных значений лучше использовать медиану —
она точнее описывает типичное время доставки. Среднее информативнее,

```python
np.random.seed(42)
delivery_simulated = np.random.normal(
    loc=30,
    scale=5,
    size=1000) #1000 наблюдений времени доставки


print(delivery_simulated.mean()) #среднее значение
print(delivery_simulated.std()) #стандартное отклонение
print(np.median(delivery_simulated)) #медиана

print("--------------------")

print(np.quantile(delivery_simulated, 0.50)) #50 процентиль
print(np.quantile(delivery_simulated, 0.90)) #90 процентиль
print(np.quantile(delivery_simulated, 0.95)) #95 процентиль
print(np.quantile(delivery_simulated, 0.99)) #99 процентиль
```

## Вывод по заданию 2.5

Высокая корреляция между продажами мороженого и несчастными случаями
на воде не доказывает причинно-следственную связь. Корреляция показывает
только то, что две величины меняются вместе, но не говорит, какая из них
является причиной.

Скрытый фактор, влияющий на оба показателя одновременно — это лето
(жаркая погода). Механизм: в жару люди покупают мороженое и одновременно
чаще идут купаться, где растёт риск несчастных случаев. Обе величины —
следствия одной общей причины.

Мой пример ложной корреляции: количество сыгранных матчей в Dota
и температура воздуха. Летом я играю чаще (больше свободного времени),
и летом температура выше. Возникает корреляция, будто Dota связана
с температурой. На самом деле общий скрытый фактор — лето: оно и даёт
больше свободного времени для игр, и повышает температуру. Прямой
причинно-следственной связи между Dota и температурой нет.

```python
x = np.array([28, 31, 29, 45, 32, 30])

print(x.mean()) #среднее значение исходной выборки

np.random.seed(42)
B = 5000
bootstrap_means = []

for i in range(B):
    sample = np.random.choice(x, size=len(x), replace=True) #новый массив из 6 чисел
    bootstrap_means.append(sample.mean()) #среднее этого массива

bootstrap_means = np.array(bootstrap_means) #переписываем в массив нампай

print(np.quantile(bootstrap_means, 0.025)) #2.5 процентиль
print(np.quantile(bootstrap_means, 0.975)) #97.5 процентиль
```

## Вопросы для анализа по заданию 3.1

1. Bootstrap использует выборку с возвращением, чтобы имитировать
случайность и получить разные виртуальные выборки. Без возвращения
все выборки были бы одинаковыми, и неопределённость оценить не удалось бы.

2. Выброс 45 расширяет распределение bootstrap-средних: когда 45 попадает
в выборку несколько раз, среднее растёт; когда не попадает — падает.
Из-за этого интервал становится шире и асимметричнее.

3. При очень маленькой выборке bootstrap неустойчив: каждая выборка
собирается из ограниченного набора чисел, и замена одного значения
сильно меняет результат.

4. Утверждение неверно. Истинное среднее фиксировано, а не случайно.
Корректная интерпретация: если строить такие интервалы многократно
на разных выборках, около 95% из них накроют истинное среднее.

## Вывод по заданию 3.2

H₀: μ_A = μ_B — среднее время доставки в двух группах одинаковое,

Ошибка I рода соответствует ложной тревоге, ошибка II рода —
пропущенному эффекту.

К прямым напрасным расходам компании ведёт ошибка I рода. Если мы

## Вывод по заданию 3.3

p-value — это вероятность увидеть такие или более экстремальные данные,

Результат статистически значим при α = 0.05, потому что 0.001 < 0.05.

В ситуации с 1 млн пользователей и эффектом 3 секунды: результат

Перед внедрением нужно учесть:

- экономическую целесообразность (3 секунды не стоят 5 млн/год);
- риски;
- качество данных;
- сегментацию пользователей.
